```python
import math

a = float(input("Nhap chieu rong a: "))
b = float(input("Nhap chieu dai b: "))

r = min(a, b) / 2

dien_tich_hcn = a * b
dien_tich_tron = 3.14 * r * r

con_lai = dien_tich_hcn - dien_tich_tron

print("Dien tich con lai:", round(con_lai, 2))
```

Bài 6

```python
import math

a = float(input("Nhap a: "))
b = float(input("Nhap b: "))
c = float(input("Nhap c: "))

if a + b > c and a + c > b and b + c > a:
    p = (a + b + c) / 2
    s = math.sqrt(p * (p - a) * (p - b) * (p - c))
    print("Dien tich tam giac:", round(s, 1))
else:
    print("Khong phai 3 canh cua tam giac")
```

```python
ten = input("Ten chu ho: ")
cu = int(input("Chi so thang truoc: "))
moi = int(input("Chi so thang nay: "))

so_dien = moi - cu
tien = 0

if so_dien <= 50:
    tien = so_dien * 1984

elif so_dien <= 100:
    tien = 50 * 1984 + (so_dien - 50) * 2050

elif so_dien <= 200:
    tien = 50 * 1984 + 50 * 2050 + (so_dien - 100) * 2380

elif so_dien <= 300:
    tien = 50 * 1984 + 50 * 2050 + 100 * 2380 + (so_dien - 200) * 2998

elif so_dien <= 400:
    tien = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + (so_dien - 300) * 3350

else:
    tien = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + 100 * 3350 + (so_dien - 400) * 3460

tien = tien * 1.08

print("Ho va ten:", ten)
print("Tien phai tra la:", int(tien))
```
